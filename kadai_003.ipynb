```python
import openpyxl
import pandas as pd
from openpyxl import Workbook
from openpyxl.utils.dataframe import dataframe_to_rows
from openpyxl.styles import PatternFill

# 「2022_年間売上表.xlsx」と「2023_年間売上表.xlsx」のSheet1のデータからデータフレームを作成
df22 = pd.read_excel('2022_年間売上表.xlsx')
df23 = pd.read_excel('2023_年間売上表.xlsx')       # ヘッダー無視

# データを連結
df = pd.concat([df22, df23],axis=0)

# 連結したデータを「商品」「売上年」ごとにグループ化し、合計金額に集約
# データを「商品」「売上年」ごとにグループ化し、売上金額の合計を計算
grouped_df = df.groupby(['商品', '売上年']).sum().reset_index()     # .reset_index(): インデックスのリセット

# 新規Excelファイル「売上集計表.xlsx」に書き込む
grouped_df.to_excel(writer, sheet_name='売上集計', index=True)

# Excelファイルの作成
wb = Workbook()
ws = wb.active

# データフレームからリストを取得してExcelファイルに追加
for r in dataframe_to_rows(grouped_df, index=False, header=True):
    ws.append(r)

# 「売上集計表.xlsx」のヘッダー部分（1行目）を薄いグレー（#F2F2F2）に設定
# ヘッダー部分の色を設定
header_fill = PatternFill(start_color='F2F2F2', end_color='F2F2F2', fill_type='solid')

for cell in ws[1]:
    cell.fill = header_fill     #セルごとに処理

# Excelファイルを保存
wb.save('売上管理表.xlsx')
```
